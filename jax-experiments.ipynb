# Jax experiments

## Vectorization

```python
import jax
import jax.numpy as jnp
```

```python
def square (x):
    return x **2

vectorized_square = jax.vmap(square)

x = jnp.array([1.0, 2.0, 3.0, 4.0])
result = vectorized_square(x)

print (result)
```

## Gradients

```python
f = lambda x: x**3 + 2*x**2 - 3*x + 1

dfdx = jax.grad(f)
jit_dfdx = jax.jit(dfdx)

print(jit_dfdx(1.0))
```
